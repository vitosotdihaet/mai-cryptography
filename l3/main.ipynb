Зададим функции для генерации текстов:
- осознанных (берем часть настоящего длинного текста)
- рандомные буквы
- рандомные слова (из настоящего текста)

```python
import random
import time

random.seed(time.time())

def random_letters(n: int, min_c: int = ord('A'), max_c: int = ord('z')) -> str:
    s = ''
    for _ in range(n):
        s += chr(random.randint(min_c, max_c))
    return s

def random_word_text(words: list[str], n: int) -> str:    
    total_length = 0
    for w in words:
        total_length += len(w)

    avg_word_len = total_length / len(words)

    t = ''
    for _ in range(int(n * 1.42 / avg_word_len)):
        t += f'{random.choice(words)} '
    return t[:n]

def random_text(text: str, n: int) -> str:
    assert len(text) >= n, 'Text\'s length is less than n'
    start = random.randint(0, len(text) - n)
    assert n == len(text[start:start + n])
    return text[start:start + n]

def get_words(text: str) -> list[str]:
    TO_REMOVE = [
        ',', '.', ';', '?', '!', '«', '»', '(', ')', '\'', '"'
    ]
    for rem in TO_REMOVE:
        text = text.replace(rem, '')
    return text.split(' ')
```

Сравнивать тексты будем двумя способами -- побуквенно и расстоянием левенштейна

```python
# pip install levenshtein
import Levenshtein

def compare_texts_by_levenshtein_distance(text1: str, text2: str) -> float:
    assert len(text1) == len(text2), 'Text\'s length do not match'
    return 1 - Levenshtein.distance(text1, text2) / len(text1)

def compare_texts_by_letters(text1: str, text2: str) -> float:
    assert len(text1) == len(text2), 'Text\'s length do not match'
    return sum(c1 == c2 for c1, c2 in zip(text1, text2)) / len(text1)
```
