Зададим функции для генерации текстов:
- осознанных (берем часть настоящего длинного текста)
- рандомные буквы
- рандомные слова (из настоящего текста)

```python
import random
import time

random.seed(time.time())

def random_letters(n: int, min_c: int = ord('A'), max_c: int = ord('z')) -> str:
    s = ''
    for _ in range(n):
        s += chr(random.randint(min_c, max_c))
    return s

def random_word_text(words: list[str], n: int) -> str:    
    total_length = 0
    for w in words:
        total_length += len(w)

    avg_word_len = total_length / len(words)

    t = ''
    for _ in range(int(n * 1.42 / avg_word_len)):
        # TODO use a random punctutaion
        t += f'{random.choice(words)} '
    return t[:n]

def random_text(text: str, n: int) -> str:
    assert len(text) >= n, 'Text\'s length is less than n'
    start = random.randint(0, len(text) - n)
    assert n == len(text[start:start + n])
    return text[start:start + n]

def get_words(text: str) -> list[str]:
    TO_REMOVE = [
        ',', '.', ';', '?', '!', '«', '»', '(', ')', '\'', '"'
    ]
    for rem in TO_REMOVE:
        text = text.replace(rem, '')
    return text.split(' ')
```

Сравнивать тексты будем двумя способами -- побуквенно и расстоянием левенштейна

```python
# pip install levenshtein
import Levenshtein

def compare_texts_by_levenshtein_distance(text1: str, text2: str) -> float:
    assert len(text1) == len(text2), 'Text\'s length do not match'
    return 1 - Levenshtein.distance(text1, text2) / len(text1)

def compare_texts_by_letters(text1: str, text2: str) -> float:
    assert len(text1) == len(text2), 'Text\'s length do not match'
    return sum(c1 == c2 for c1, c2 in zip(text1, text2)) / len(text1)
```

```python
from collections import defaultdict
import itertools

INIT_TEXT_LENGTH = 100
DELTA_TEXT_LENGTH = 50
MAX_TEXT_LENGTH = 1000 + DELTA_TEXT_LENGTH

TEST_COUNT = 5

words = []
for t in real_texts:
    words.extend(get_words(t))

print(f'An average of {TEST_COUNT} tests of the same length\n')
for text_length in range(INIT_TEXT_LENGTH, MAX_TEXT_LENGTH, DELTA_TEXT_LENGTH):
    similarities = defaultdict(lambda: {'letters': 0., 'levenshtein': 0.})

    print(f'{f"{text_length}:":<5} {"Pair":<17} {"By letters":<15} {"Levenshtein":<10}')
    print("-" * 50)

    for _ in range(TEST_COUNT):
        real_text_1 = random_text(random.choice(real_texts), text_length)
        real_text_2 = random_text(random.choice(real_texts), text_length)
        words_text = random_word_text(words, text_length)
        letters_text = random_letters(text_length, ord('А'), ord('я'))

        sources = [
            ('Real 1', real_text_1),
            ('Real 2', real_text_2),
            ('Words', words_text),
            ('Letters', letters_text),
        ]

        pairs = itertools.combinations(sources, 2)


        for (name1, t1), (name2, t2) in pairs:
            by_letters = compare_texts_by_letters(t1, t2)
            by_levenshtein = compare_texts_by_levenshtein_distance(t1, t2)
            key = tuple(sorted((name1, name2)))
            similarities[key]['letters'] += by_letters
            similarities[key]['levenshtein'] += by_levenshtein

    for (name1, name2), scores in similarities.items():
        avg_letters = scores['letters'] / TEST_COUNT
        avg_levenshtein = scores['levenshtein'] / TEST_COUNT
        print(f'{"":<5} {name1:<8}: {name2:<7} {avg_letters:<15.10f} {avg_levenshtein:<10.10f}')

    print()
```
